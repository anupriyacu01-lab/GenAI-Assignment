### Writing your OpenRouter code for the OpenAI model


### Please Write your answer here

Delete the `raise NotImplementedError()` when you start writing your code.
Use OpenRouter and the `OPENAI_MODEL` model variable.


````python
# --- Student TODO: implement only this function. Keep it minimal. ---
from openai import OpenAI

def call_openai(prompt: str) -> str:
    """
    Return the plain text output from the selected OpenAI model through OpenRouter.
    Use OPENROUTER_API_KEY, OPENROUTER_BASE_URL, and OPENAI_MODEL.
    """
    client = OpenAI(
        api_key=OPENROUTER_API_KEY,
        base_url=OPENROUTER_BASE_URL,
    )
    response = client.chat.completions.create(
        model=OPENAI_MODEL,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3,
        max_tokens=250,
    )
    return response.choices[0].message.content or ""
````

Reflection on Multibot Chat Playground
The objective of this assignment was to build a Multibot Chat Playground using OpenRouter and compare the performance of multiple Large Language Models (LLMs) on different types of tasks. I selected two models available through OpenRouter: OpenAI GPT-4o Mini and Google Gemini 2.0 Flash. Each model received the same prompts for three categories of tasks: creative writing, analytical reasoning, and instruction following. This experiment helped me understand how different models respond to identical requests and highlighted the strengths and limitations of each model.

For the creative writing task, the prompt asked the models to generate a short science-fiction story about a college student discovering an AI system capable of predicting future events. Both models successfully generated complete stories, but their styles differed noticeably. GPT-4o Mini produced a more engaging and detailed story with a stronger narrative flow and a surprising ending. Gemini 2.0 Flash generated an imaginative story with interesting ideas and a clear structure, but its storytelling was slightly less detailed. Based on creativity, originality, and overall storytelling quality, GPT-4o Mini performed better in this category.

For the analytical reasoning task, the models were asked to determine the best combination of projects under a budget constraint. Both models correctly identified that Projects A and C should be selected because their combined cost is $90,000 and their combined profit is $140,000, which is the highest profit achievable while remaining within the budget. GPT-4o Mini provided a detailed step-by-step explanation showing each possible combination and its profitability. Gemini 2.0 Flash also reached the correct answer but used a more concise explanation. Both models demonstrated strong reasoning capabilities, but GPT-4o Mini provided slightly clearer justification for its decision.

For the instruction-following task, the models were required to return a JSON object exactly as specified and without any additional text. GPT-4o Mini and Gemini 2.0 Flash both successfully returned the requested JSON structure. However, GPT-4o Mini followed the formatting instructions more consistently and precisely. This task demonstrated the importance of instruction-following ability when building applications that require structured outputs such as APIs, automation workflows, and data-processing systems.

Response speed was another factor observed during testing. Gemini 2.0 Flash generally generated responses faster than GPT-4o Mini. This makes Gemini a strong choice for applications where low latency is important, such as chatbots and real-time assistants. GPT-4o Mini was slightly slower but consistently produced more detailed and polished outputs. The experiment highlighted a common trade-off in AI systems: faster models may sacrifice some depth, while more advanced models may require additional processing time.

Several API mechanics influenced the results. OpenRouter provides a unified API endpoint that allows developers to switch between different AI providers without changing the overall application structure. Model selection plays a major role because each model has different capabilities and optimization goals. Parameters such as temperature, max tokens, and response length can also affect output quality. Lower temperature values generally improve consistency, while higher values encourage creativity and diversity. Since the prompts and settings remained the same for both models, the observed differences were primarily due to the models themselves.

A practical production system could automatically route tasks to the most suitable model. For example, creative writing tasks could be routed to GPT-4o Mini because of its stronger storytelling ability. Tasks requiring quick responses could be routed to Gemini 2.0 Flash due to its speed advantage. Instruction-following tasks involving structured output could be assigned to whichever model demonstrates the highest compliance during testing. Such a routing system would improve efficiency while maximizing output quality.

This experiment demonstrates both the strengths and limitations of using multiple AI models within a single application. One major advantage is flexibility, as developers can choose the most appropriate model for a specific task. Another advantage is reliability, because one provider can serve as a backup if another becomes unavailable. However, there are also challenges. Different models may produce different styles, levels of detail, and formatting behavior. Managing multiple providers can increase application complexity and require additional testing and monitoring.

Overall, GPT-4o Mini performed best in creative writing and analytical reasoning, while Gemini 2.0 Flash demonstrated excellent response speed and reliable performance. The results show that no single model is ideal for every situation. Instead, the best choice depends on the specific requirements of the task. OpenRouter makes it easy to compare and integrate multiple models, enabling developers to build more flexible, efficient, and capable AI-powered applications.
